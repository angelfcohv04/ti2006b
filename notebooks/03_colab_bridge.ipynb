**Pregunta:** ¿Qué categoría de gastos tiene el mayor gasto promedio en este ejemplo?

**Procedencia de datos:** datos completamente sintéticos creados en este notebook. No representan personas ni transacciones reales.

**Resultado esperado:** una tabla con el número de operaciones y el gasto promedio por categoría, seguida de una conclusión reproducible.

**Entorno:** Python 3. Se utiliza exclusivamente la biblioteca estándar de Python, por lo que no se requiere instalar paquetes adicionales. Ejecuta todas las celdas en orden desde una sesión nueva.

```python
import sys
from statistics import mean

print("Versión de Python:", sys.version)
print("Dependencias: biblioteca estándar de Python; no se requieren instalaciones.")
```

```python
# Datos ficticios inventados para esta actividad (montos en pesos).
transacciones = [
    {"categoria": "Alimentos", "monto": 120},
    {"categoria": "Alimentos", "monto": 180},
    {"categoria": "Transporte", "monto": 45},
    {"categoria": "Transporte", "monto": 75},
    {"categoria": "Entretenimiento", "monto": 210},
    {"categoria": "Entretenimiento", "monto": 150},
]
print("Transacciones sintéticas:", len(transacciones))
for transaccion in transacciones:
    print(transaccion)
```

```python
# Agrupar los importes y calcular la media para cada categoría.
grupos = {}
for transaccion in transacciones:
    grupos.setdefault(transaccion["categoria"], []).append(transaccion["monto"])

resumen = {categoria: {"cantidad": len(montos), "promedio": mean(montos)}
           for categoria, montos in grupos.items()}
print(f"{'Categoría':<20} {'Operaciones':>11} {'Promedio':>10}")
for categoria, datos in sorted(resumen.items()):
    print(f"{categoria:<20} {datos['cantidad']:>11} ${datos['promedio']:>9.2f}")

mayor = max(resumen, key=lambda categoria: resumen[categoria]["promedio"])
print(f"Mayor gasto promedio: {mayor} (${resumen[mayor]['promedio']:.2f}).")
```

## Conclusión

En estos datos sintéticos, entretenimiento tiene el gasto promedio más alto: 180 pesos por operación. Alimentos tiene un promedio de 150 pesos y transporte de 60 pesos. El ejemplo es pequeño y ficticio, por lo que no permite sacar conclusiones sobre hábitos reales.
